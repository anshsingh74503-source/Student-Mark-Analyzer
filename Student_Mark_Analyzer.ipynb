```python
import pandas as pd

data = {
    "Roll_No": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Aman", "Rahul", "Priya", "Anjali", "Rohit",
             "Sneha", "Vikas", "Neha", "Karan", "Pooja"],
    "English": [85, 67, 92, 45, 78, 88, 56, 95, 72, 81],
    "Maths": [90, 72, 95, 38, 65, 91, 48, 98, 75, 84],
    "Science": [88, 75, 89, 42, 70, 85, 52, 96, 68, 87]
}

df = pd.DataFrame(data)

print("STUDENT MARKS DATASET")
display(df)
```

```python
# Calculate Total Marks
df["Total"] = df["English"] + df["Maths"] + df["Science"]

# Calculate Percentage
df["Percentage"] = (df["Total"] / 300) * 100

# Grade Calculation
def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

df["Grade"] = df["Percentage"].apply(calculate_grade)

print("STUDENT MARKS ANALYSIS")
display(df)
```

```python
# Pass or Fail Status
df["Result"] = df.apply(
    lambda row: "Pass" if row["English"] >= 35
    and row["Maths"] >= 35
    and row["Science"] >= 35 else "Fail",
    axis=1
)

# Find Topper
topper = df.loc[df["Total"].idxmax()]

print("STUDENT RESULT ANALYSIS")
display(df)

print("\nTOPPER DETAILS")
print("Name:", topper["Name"])
print("Total Marks:", topper["Total"])
print("Percentage:", round(topper["Percentage"], 2))
print("Grade:", topper["Grade"])
```

```python
import matplotlib.pyplot as plt

# Graph 1: Total Marks
plt.figure(figsize=(10, 5))
plt.bar(df["Name"], df["Total"])
plt.title("Student Total Marks")
plt.xlabel("Students")
plt.ylabel("Total Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Graph 2: Percentage
plt.figure(figsize=(10, 5))
plt.bar(df["Name"], df["Percentage"])
plt.title("Student Percentage")
plt.xlabel("Students")
plt.ylabel("Percentage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Graph 3: Grade Distribution
grade_counts = df["Grade"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(grade_counts, labels=grade_counts.index, autopct="%1.1f%%")
plt.title("Grade Distribution")
plt.show()
```

```python
# Save Student Marks Data
df.to_csv("student_marks.csv", index=False)

# Save Graph as Image
plt.figure(figsize=(10, 5))
plt.bar(df["Name"], df["Total"])
plt.title("Student Marks Report")
plt.xlabel("Student Name")
plt.ylabel("Total Marks")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("student_marks_report.png")

print("CSV file and Graph saved successfully!")
```
